## Result and recommendation

The context branch is promising in this paired smoke test: people-proposal frames increased from 17 to 19, and vehicle-proposal frames from 0 to 15, while the animal decisions remained identical. Both negative controls stayed closed (0/41 sampled frames). Keep this as an experimental branch pending annotated aerial co-occurrence tests.

The reviewers still produced isolated false animal confirmations on the controls: beach crop at 9 seconds, beach scene at 3 seconds, and rocky scene at about 9 seconds. Agreement/temporal requirements blocked these from opening the display. This is a useful regression result, not proof hallucinations are solved. Vehicle proposals sometimes overlap the same truck; counts must not be interpreted as unique vehicles.

## Fixed paired design

Baseline: DMD animal and context proposals. Combined: exactly the same DMD animal proposals, with YOLO-World full-frame person/vehicle context. Both share identical actual Gemini animal judgments. This isolates the context branch, rather than confounding it with separate stochastic Gemini answers.

Every sixth 2-FPS sample is reviewed: at most three native animal crops plus overview, and a separate independent scene check. The unchanged gate accepts same-check crop+scene agreement or two consecutive scene confirmations within four seconds; recent evidence is held for three seconds. Both reviewers use the same Gemini model and may make correlated mistakes.

Only source seconds 12–24 were sampled at 2 FPS, with the window selected before model scoring to include visible co-occurrence.

The native 18-second frame was manually screened before inference and shows the seal, responders and a pickup. Derived previews are sampled, resized and overlaid; no endorsement is implied.

## Remaining limits

This is an offline research pipeline, not app integration or live-stream validation. Source-time replay excludes inference/API scheduling delay. No injury, urgency, danger, harassment or rescue-need inference is performed. Natural shot cuts, current animal exits and display of every individual context box remain unvalidated. A three-second evidence hold is not proof the animal remains visible on every frame.

The independent scene checker can hallucinate animals, and both reviewers share the same model. The two negatives are repeated-call regressions rather than held-out negatives. This test alone cannot justify deployment.
